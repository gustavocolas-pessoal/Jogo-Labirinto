```python

pg.init() # inicializa o Pygame

largura = 900
altura = 600

tela = pg.display.set_mode((largura, altura)) # cria a janela com as dimensoes definidas
pg.display.set_caption("LABIRINTO") # define o titulo da janela

# define as tres fontes de texto em tamanhos distintos
arial = pg.font.SysFont("Arial", 30)
arial_pequeno = pg.font.SysFont("Arial", 22)
arial_huge = pg.font.SysFont("Arial", 50)

# definindo as cores (tuplas RGB)
branco = (230,230,230)
preto = (0,0,0)
quase_preto = (30,30,30)
verdin = (40,147,47)
vermelhin = (150,3,3)
bege = (194,176,146)
cinza = (50,50,50)
azulzin_1 = (135, 206, 235)
azulzin_2 = (50,150,250)

labirinto = pg.Rect(50, 50, 400, 400)
coords = [50,100,150,200,250,300,350,400,450]


def desenha_grade():
    pg.draw.rect(tela, azulzin_1, labirinto)
    for x in coords:
        for y in coords:
            grid_h = pg.Rect(x, y, 50, 3) 
            grid_v = pg.Rect(x, y, 3, 53) 
            if x != 450:
                pg.draw.rect(tela, bege, grid_h)
            if y != 450:
                pg.draw.rect(tela, bege, grid_v)


def desenha_paredes(qual):

    for ph in qual[0]:
        x = coords[ph[0]]
        y = coords[ph[1]]
        parede = (x,y,50,4)
        pg.draw.rect(tela,preto,parede)
    for pv in qual[1]:
        x = coords[pv[0]]
        y = coords[pv[1]]
        parede = (x,y,4,51)
        pg.draw.rect(tela,preto,parede)

def desenha_pos_atual(pos_atual):
    x = 65 + (pos_atual[0])*50
    y = 65 + (pos_atual[1])*50
    onde = pg.Rect(x,y,25,25)
    pg.draw.rect(tela,vermelhin,onde)

rodando = True
grade = 0
desenhar_labirinto = False
lab = 0
n = 0
pos_atual = (0,0) 

while rodando:

    for event in pg.event.get():

        if event.type == pg.QUIT:
            pg.quit()
            sys.exit()

        elif event.type == pg.KEYDOWN:
            if event.key == pg.K_SPACE:
                rodando = False

            elif event.key == pg.K_r:
                n += 1
                qual_labirinto = lista_labirintos[(n%2)]

            elif event.key == pg.K_l:
                desenhar_labirinto += 1 

            elif event.key == pg.K_UP:
                permite_movimento(tenta_vertical_entrada())

            elif event.key == pg.K_DOWN:
                permite_movimento(tenta_vertical_saida())

            elif event.key == pg.K_RIGHT:
                permite_movimento(tenta_horizontal_saida())

            elif event.key == pg.K_LEFT:
                permite_movimento(tenta_horizontal_entrada())

    # Limpa a tela
    tela.fill(branco)
    desenha_grade()
    desenha_pos_atual(pos_atual)

    if desenhar_labirinto % 2 == 0:
        desenha_paredes(qual_labirinto)

    # Atualiza a janela
    pg.display.flip() 

pg.quit()
sys.exit()
```
